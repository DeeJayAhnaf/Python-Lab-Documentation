```python
class Car:
    def __init__(self, make, model):
        self.make = make
        self.model = model

car = Car("Honda", "Civic")

class Dog:
    def __init__(self, name):
        self.name = name
    def bark(self):
        return f"{self.name} barks"

dog = Dog("Rex")

class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance
    def deposit(self, amt):
        self.balance += amt

acc = BankAccount(100)
acc.deposit(50)

class Employee:
    company = "TechCorp"
    def __init__(self, name):
        self.name = name

emp = Employee("Sarah")

class Person:
    def __init__(self, age):
        self.__age = age
    def get_age(self):
        return self.__age

p = Person(25)

class Animal:
    def speak(self):
        return "Sound"

class Cat(Animal):
    def speak(self):
        return "Meow"

c = Cat()

class Book:
    def __init__(self, pages):
        self.pages = pages
    def __len__(self):
        return self.pages

b = Book(200)

print(car.make)
print(dog.bark())
print(acc.balance)
print(emp.company)
print(p.get_age())
print(c.speak())
print(len(b))
```
