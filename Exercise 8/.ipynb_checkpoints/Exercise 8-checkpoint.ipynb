```python
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def make_sound(self):
        return f"{self.name} barks"

class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, doors):
        super().__init__(brand)
        self.doors = doors

class Shape:
    def area(self):
        return 0

class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side ** 2

class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, emp_id):
        super().__init__(name)
        self.emp_id = emp_id

class Developer(Employee):
    def __init__(self, name, emp_id, lang):
        super().__init__(name, emp_id)
        self.lang = lang

class Writer:
    def write(self):
        return "Writing..."

class Speaker:
    def speak(self):
        return "Speaking..."

class Author(Writer, Speaker):
    pass

dog = Dog("Buddy")
car = Car("Ford", 4)
sq = Square(4)
dev = Developer("Bob", 201, "Python")
author = Author()

print(dog.make_sound())
print(car.brand, car.doors)
print(sq.area())
print(dev.name, dev.emp_id, dev.lang)
print(author.write(), author.speak())
```
