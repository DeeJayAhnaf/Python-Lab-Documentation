```python
def personalize_greeting(name, role):
    print(f"Hello {name}, your role is {role}.")

personalize_greeting("Alice", "Developer")
```

```python
def calculate_square(number):
    return number * number

squared_val = calculate_square(5)
print(f"5 squared is {squared_val}")
```

```python
def power(base, exponent=2):
    return base ** exponent

print(f"4^2 = {power(4)}")
print(f"4^3 = {power(4, 3)}")
```

```python
def get_min_max(numbers):
    return min(numbers), max(numbers)

scores = [45, 88, 12, 99, 63]
min_score, max_score = get_min_max(scores)
print(f"Min = {min_score}, Max = {max_score}")
```

```python
def summarize_data(*args, **kwargs):
    print(f"Positional args: {args}")
    print(f"Keyword args: {kwargs}")

summarize_data(10, 20, 30, user="Bob", status="Active")
```

```python
def outer_calculator(x):
    factor = 2

    def inner_multiplier():
        return x * factor

    return inner_multiplier()

print(f"Result: {outer_calculator(10)}")
```
