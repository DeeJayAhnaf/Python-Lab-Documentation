```python
class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c

calc = Calculator()
print(calc.add(10))
print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```

```python
class MathOperations:
    def multiply(self, *args):
        if not args:
            return 0
        result = 1
        for num in args:
            result *= num
        return result

ops = MathOperations()
print(ops.multiply(5))
print(ops.multiply(2, 3, 4))
```

```python
class Formatter:
    def format_input(self, value):
        if isinstance(value, int):
            return f"Integer: {value * 2}"
        elif isinstance(value, str):
            return f"String: {value.upper()}"
        elif isinstance(value, list):
            return f"List length: {len(value)}"
        return "Unsupported type"

fmt = Formatter()
print(fmt.format_input(10))
print(fmt.format_input("hello"))
print(fmt.format_input([1, 2, 3]))
```
