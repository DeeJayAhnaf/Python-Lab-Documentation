```python
import re

text = "The user contact email is alex.dev@example.com."
pattern = r"alex\.dev@example\.com"

match = re.search(pattern, text)
if match:
    print(match.group())
    print(match.span())
```

```python
import re

text = "Orders: #1001, #1002, #1003, and #1004"
pattern = r"#\d+"

matches_list = re.findall(pattern, text)
print(matches_list)

for match in re.finditer(pattern, text):
    print(f"Found {match.group()} at index {match.start()}:{match.end()}")
```

```python
import re

username_pattern = r"^[a-zA-Z0-9_]{3,16}$"

valid_user = "user_123"
invalid_user = "usr@99!"

print(bool(re.fullmatch(username_pattern, valid_user)))
print(bool(re.fullmatch(username_pattern, invalid_user)))
```

```python
import re

log_entry = "2026-09-27 18:00:00 [ERROR] Connection timed out"
pattern = r"(?P<date>\d{4}-\d{2}-\d{2})\s+(?P<time>\d{2}:\d{2}:\d{2})\s+\[(?P<level>[A-Z]+)\]\s+(?P<message>.*)"

match = re.search(pattern, log_entry)
if match:
    print(match.groupdict())
    print(match.group("level"))
    print(match.group("message"))
```
