```python
dirty_text = "   hello world   \n"

cleaned_both = dirty_text.strip()
cleaned_left = dirty_text.lstrip()
cleaned_right = dirty_text.rstrip()

print(f"'{cleaned_both}'")
print(f"'{cleaned_left}'")
print(f"'{cleaned_right}'")
```
