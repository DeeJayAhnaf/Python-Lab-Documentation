```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['Alice', 'Bob'],
    'Score': [85, 90]
})

df2 = pd.DataFrame({
    'ID': [3, 4],
    'Name': ['Charlie', 'David'],
    'Score': [78, 92]
})

df_concat_rows = pd.concat([df1, df2], ignore_index=True)
print(df_concat_rows)
```

```python
import pandas as pd

df_left = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})

df_right = pd.DataFrame({
    'Age': [25, 30, 35],
    'City': ['New York', 'London', 'Paris']
})

df_concat_cols = pd.concat([df_left, df_right], axis=1)
print(df_concat_cols)
```

```python
import pandas as pd

df_employees = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Dept_ID': [1, 2, 1, 3]
})

df_departments = pd.DataFrame({
    'Dept_ID': [1, 2, 4],
    'Dept_Name': ['HR', 'Engineering', 'Marketing']
})

inner_join = pd.merge(df_employees, df_departments, on='Dept_ID', how='inner')
left_join = pd.merge(df_employees, df_departments, on='Dept_ID', how='left')
right_join = pd.merge(df_employees, df_departments, on='Dept_ID', how='right')
outer_join = pd.merge(df_employees, df_departments, on='Dept_ID', how='outer')

print(inner_join)
print(left_join)
print(outer_join)
```
