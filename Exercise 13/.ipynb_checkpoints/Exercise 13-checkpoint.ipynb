```python
import pandas as pd

df = pd.DataFrame({
    'Store': ['North', 'North', 'South', 'South', 'North'],
    'Product': ['A', 'B', 'A', 'B', 'A'],
    'Sales': [200, 150, 300, 400, 250]
})

store_summary = df.groupby('Store').agg(
    total_sales=('Sales', 'sum'),
    avg_sales=('Sales', 'mean')
)

df['Store_Max'] = df.groupby('Store')['Sales'].transform('max')
df['Pct_Of_Max'] = df['Sales'] / df['Store_Max']

active_stores = df.groupby('Store').filter(lambda x: x['Sales'].sum() > 500)

print(store_summary)
print(df)
print(active_stores)
```
