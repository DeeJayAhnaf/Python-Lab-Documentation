```python
empty_tuple = ()
single_element_tuple = (5,)
numbers_tuple = (10, 20, 30, 40, 50)
mixed_tuple = (1, "Hello", 3.14, True)
tuple_without_parentheses = 100, 200, 300

print(type(empty_tuple))
print(single_element_tuple)
print(numbers_tuple)
print(mixed_tuple)
print(tuple_without_parentheses)
```

```python
data = ("apple", "banana", "cherry", "date", "elderberry")

first_item = data[0]
last_item = data[-1]
slice_middle = data[1:4]
slice_step = data[::2]
reversed_tuple = data[::-1]

print(first_item)
print(last_item)
print(slice_middle)
print(slice_step)
print(reversed_tuple)
```

```python
coordinates = (10, 20, 30)
x, y, z = coordinates

values = (1, 2, 3, 4, 5)
head, *middle, tail = values

print(f"x: {x}, y: {y}, z: {z}")
print(f"head: {head}, middle: {middle}, tail: {tail}")
```

```python
tuple_a = (1, 2, 3)
tuple_b = (4, 5, 6)

combined_tuple = tuple_a + tuple_b
repeated_tuple = tuple_a * 3

print(combined_tuple)
print(repeated_tuple)
```

```python
numbers = (10, 20, 30, 20, 40, 20, 50)

count_20 = numbers.count(20)
index_30 = numbers.index(30)
exists = 40 in numbers

print(count_20)
print(index_30)
print(exists)
```

```python
scores = (45, 88, 12, 99, 63)

tuple_length = len(scores)
min_value = min(scores)
max_value = max(scores)
total_sum = sum(scores)
sorted_list = sorted(scores)

print(tuple_length)
print(min_value)
print(max_value)
print(total_sum)
print(sorted_list)
```
