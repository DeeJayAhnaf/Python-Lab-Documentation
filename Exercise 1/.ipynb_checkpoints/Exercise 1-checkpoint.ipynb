```python
age = int(input('Enter the Age : '))

if age >= 18:
    print("You are an adult.")
```

```python
score = int(input('Enter the Score : '))

if score >= 50:
    print("Pass")
else:
    print("Fail")
```

```python
marks = int(input('Enter the Marks : '))

if marks >= 90:
    print("Grade: A+")
elif marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
else:
    print("Grade: C or below")
```

```python
num = int(input('Enter the Number : '))

if num > 0:
    if num % 2 == 0:
        print("Positive Even Number")
    else:
        print("Positive Odd Number")
else:
    print("Zero or Negative Number")
```

```python
# Iterating through a range of numbers (0 to 4)
for i in range(5):
    print(f"Number: {i}")

# Iterating over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# Print a 3x3 grid / coordinate map
for row in range(1, 4):
    for col in range(1, 4):
        print(f"({row}, {col})", end=" ")
    print()  # Newline after each row
```

```python
count = 5

while count > 0:
    print(f"Countdown: {count}")
    count -= 1

print("Loop has been Executed Successfully!")
```

```python
# Find the first even number greater than 10
number = 1

while True:
    number += 1
    if number <= 10:
        continue  # Skip numbers 1 through 10
    
    if number % 2 == 0:
        print(f"Found target number: {number}")
        break  # Exit loop completely
```

```python
outer_count = 1

while outer_count <= 3:
    inner_count = 1
    while inner_count <= 2:
        print(f"Outer: {outer_count}, Inner: {inner_count}")
        inner_count += 1
    outer_count += 1
```
