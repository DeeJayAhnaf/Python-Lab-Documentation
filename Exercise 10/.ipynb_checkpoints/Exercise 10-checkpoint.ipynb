```python
with open("example.txt", "w", encoding="utf-8") as f:
    f.write("Hello, World!\n")
    f.writelines(["Line 2: Python file I/O\n", "Line 3: Text mode processing\n"])

with open("example.txt", "r", encoding="utf-8") as f:
    content = f.read()

print(content)

with open("example.txt", "a", encoding="utf-8") as f:
    f.write("Line 4: Appended text\n")

with open("example.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()

print(lines)
```

```python
with open("example.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(line.strip())
```

```python
binary_data = bytes([65, 66, 67, 68, 69, 70])

with open("data.bin", "wb") as f:
    f.write(binary_data)

with open("data.bin", "rb") as f:
    content = f.read()

print(content)
print(list(content))
```
